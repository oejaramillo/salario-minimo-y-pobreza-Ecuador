## Ingreso descontando cargas familiares

Utilizando la metodología del autor dividimos el ingreso del hogar para la escala $(A_{i}+kC_{i})^{s}$ donde $A_{i}$ es al número de adultos, $C_{i}$ es el número de niños en el hogar $i$. $k$ es el costo en recursos de cada niño y $s$ busca reflejar las restricciones

```python
k = 0.4
s = 0.9
```

```python
# Si es necesario calcular el número de niños
data['es_nino'] = data['edad'] < 10

data['ninos'] = data.groupby('identificador hogares')['es_nino'].transform('sum')

# Si es necesario calcular el número de adultos
data['es_adulto'] = data['edad'] > 10

data['adultos'] = data.groupby('identificador hogares')['es_adulto'].transform('sum')




# Si es necesario calcular el número de niños
data2['es_nino'] = data2['edad'] < 10

data2['ninos'] = data2.groupby('idhogar')['es_nino'].transform('sum')

# Si es necesario calcular el número de adultos
data2['es_adulto'] = data2['edad'] > 10

data2['adultos'] = data2.groupby('idhogar')['es_adulto'].transform('sum')




# Si es necesario calcular el número de niños
data3['es_nino'] = data3['edad'] < 10

data3['ninos'] = data3.groupby('idhogar1')['es_nino'].transform('sum')

# Si es necesario calcular el número de adultos
data3['es_adulto'] = data3['edad'] > 10

data3['adultos'] = data3.groupby('idhogar1')['es_adulto'].transform('sum')




# Si es necesario calcular el número de niños
data4['es_nino'] = data4['edad'] < 10

data4['ninos'] = data4.groupby('identificador hogares')['es_nino'].transform('sum')

# Si es necesario calcular el número de adultos
data4['es_adulto'] = data4['edad'] > 10

data4['adultos'] = data4.groupby('identificador hogares')['es_adulto'].transform('sum')
```

```python
data['escala'] = (data['adultos'] + k * data['ninos']) ** s

data2['escala'] = (data2['adultos'] + k * data2['ninos']) ** s

data3['escala'] = (data3['adultos'] + k * data3['ninos']) ** s

data4['escala'] = (data4['adultos'] + k * data4['ninos']) ** s
```

```python
data['ingr_t'] = data['ingr_h'] / data['escala']

data2['ingr_t'] = data2['ingr_h'] / data2['escala']

data3['ingr_t'] = data3['ingr_h'] / data3['escala']

data4['ingr_t'] = data4['ingr_h'] / data4['escala']
```

## Cálculo del índice de pobreza de Foster, Greer y Thorbecke

Para calcular un ínidce de pobreza se utiliza a Foster, Greer y Thorbecke (1984), ya que satisface algunas caracterísitcas de distribución que son positivas e igual a las enunciadas por Sen, el autor usa el mismo índice.

$FGT_{\alpha} = \frac{1}{N}\sum_{i=1}^{H}\left(\frac{z-y_{i}}{z}\right)^{\alpha} $

Donde $z$ es el umbral de pobreza, $N$ es el número de personas en la economía, $H$ es el número de pobres (personas debajo de la línea de pobreza) $y_{i}$ es el ingreso de cada individuo. Mientras mayor es el valor de $\alpha$ mayor es el peso de los individuos más pobres, mayor $FGT$ mayor pobreza en la economía.

```python
# Buscamos a los que están por debajo del umbral de pobreza
data['pobres'] = (data['ingr_t'] - data['umbral']) < 0

# Cantidad de pobres por región
data['pobres_region'] = data.groupby('rn')['pobres'].transform('sum')

# Cantidad de personas por región
data['identificador persona'] = data[['rn', 'dominio', 'ciudad', 'zona', 'sector', 'vivienda', 'hogar', 'formul', 'persona']].astype(str).agg('-'.join, axis=1)
data['ingr_h'] = data.groupby('identificador hogares')['ingr_r'].transform('sum')
data['personas'] = data.groupby('rn')['identificador persona'].transform('sum')




# Buscamos a los que están por debajo del umbral de pobreza
data2['pobres'] = (data2['ingr_t'] - data2['umbral']) < 0

# Cantidad de pobres por región
data2['pobres_region'] = data2.groupby('rn')['pobres'].transform('sum')

# Cantidad de personas por región
data2['ingr_h'] = data2.groupby('idhogar')['ingr_r'].transform('sum')
data2['personas'] = data2.groupby('rn')['id_persona'].transform('sum')




# Buscamos a los que están por debajo del umbral de pobreza
data3['pobres'] = (data3['ingr_t'] - data3['umbral']) < 0

# Cantidad de pobres por región
data3['pobres_region'] = data3.groupby('rn')['pobres'].transform('sum')

# Cantidad de personas por región
data3['identificador persona'] = data3[['idhogar1', 'persona']].astype(str).agg('-'.join, axis=1)
data3['ingr_h'] = data3.groupby('idhogar1')['ingr_r'].transform('sum')
data3['personas'] = data3.groupby('rn')['identificador persona'].transform('sum')




# Buscamos a los que están por debajo del umbral de pobreza
data4['pobres'] = (data4['ingr_t'] - data4['umbral']) < 0

# Total de personas pobres (a nivel nacional o del dataset completo)
total_pobres = data4['pobres'].sum()
print("Total de personas pobres:", total_pobres)

# Cantidad de personas por región
data4['identificador persona'] = data4[['identificador hogares', 'persona']].astype(str).agg('-'.join, axis=1)
data4['ingr_h'] = data4.groupby('identificador hogares')['ingr_r'].transform('sum')

# Total de personas (únicas) en el dataset
total_personas = data4['identificador persona'].nunique()
```

## Calculo del índice de desigualdad de Atkinson

Vamos a calcular el índice de desigualdad de atkinson con un parámetro $\epsilon$ de aversión a la desigualdad y un $\mu$ que es igual a la media de los ingresos individuales, con la siguiente fórmula.

$A = 1-\frac{1}{\mu}\left(\frac{1}{N}\sum_{i=1}^{N}y^{1-\epsilon}\right)^{1/(1-\epsilon}$

Donde $y_{i}$ es el ingreso individual y $\mu$ es el ingreso medio

```python
# Ingreso promedio por región
data['mu'] = data.groupby('rn')['ingr_t'].transform('mean')

# Ingreso promedio por región
data2['mu'] = data2.groupby('rn')['ingr_t'].transform('mean')

# Ingreso promedio por región
data3['mu'] = data3.groupby('rn')['ingr_t'].transform('mean')

# Ingreso promedio por región
mu = data4['ingr_t'].mean()
data4['mu'] = mu
```
